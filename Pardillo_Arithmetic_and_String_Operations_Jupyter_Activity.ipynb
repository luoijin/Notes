## 14. Circle Calculator

Given `radius = 7`, calculate the area and circumference of a circle. Use:

```python

```

Formulas:
- Area = π × r²
- Circumference = 2 × π × r

````python
radius = 7
pi = 3.14159

Area = pi * radius ** 2
Circumference = 2 * pi * radius
````

````python
print(Area)
print(Circumference)
````

## 15. Salary Calculator

An employee works `40` hours per week and earns ₱150 per hour. Calculate:
- Weekly salary
- Monthly salary, assuming 4 weeks per month
- Annual salary, assuming 12 months per year

````python
hours_per_week = 40
hourly_rate = 150
````

````python
weekly_salary =  hours_per_week * hourly_rate
monthly_salary = (4 * hours_per_week) * hourly_rate
annual_salary = (12 * hours_per_week) * hourly_rate
````

````python
print(weekly_salary)
print(monthly_salary)
print(annual_salary)
````
